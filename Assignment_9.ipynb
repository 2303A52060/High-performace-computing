Assignment 2: Integer Data Exchange between Multiple MPI
Processes

Scenario:

A parallel sensor network distributes integer readings from a
control node to compute nodes.
Objective

Use MPI Send and Receive for integer data communication among
multiple processes.

Tasks
1. Run the MPI program with at least 4
processes.

2. Let process rank 0 send different integer values to ranks 1,
2, and 3.

3. Each receiving process prints the received value along with
its rank.

4. Ensure correct use of tags to differentiate messages


```python
import sys
!{sys.executable} -m pip install mpi4py

from mpi4py import MPI

# Initialize MPI
comm = MPI.COMM_WORLD
rank = comm.Get_rank()
size = comm.Get_size()

# Ensure at least 4 processes
if size < 4:
    if rank == 0:
        print(f"Please run with at least 4 processes for this MPI code to function correctly. Current number of processes: {size}")
    # Do not attempt MPI communication if size is insufficient
    # This 'fix' prevents the MPI_ERR_RANK error in a single-process environment.
else:
    # Process 0 sends data
    if rank == 0:
        data1 = 10
        data2 = 20
        data3 = 30

        print(f"Process {rank} sending data...")

        # Sending with different tags
        comm.send(data1, dest=1, tag=101)
        comm.send(data2, dest=2, tag=102)
        comm.send(data3, dest=3, tag=103)

    # Receiving processes
    elif rank == 1:
        value = comm.recv(source=0, tag=101)
        print(f"Process {rank} received value: {value}")

    elif rank == 2:
        value = comm.recv(source=0, tag=102)
        print(f"Process {rank} received value: {value}")

    elif rank == 3:
        value = comm.recv(source=0, tag=103)
        print(f"Process {rank} received value: {value}")

    # Extra processes (if more than 4)
    else:
        print(f"Process {rank} has no data to receive")
```

Learning Outcomes

• Point-to-point communication

• Role of ranks in MPI


Message tagging: The code demonstrates message tagging by using different tag values (101, 102, 103) for the comm.send and comm.recv operations, ensuring that specific data is received by the intended process.

Multiple send/receive operations: Process 0 performs multiple comm.send operations, and processes 1, 2, and 3 each perform comm.recv operations, showcasing multiple communication instances.

Process coordination in MPI: The code explicitly coordinates processes by having Process 0 send data to specific ranks (1, 2, 3), and these ranks then wait to receive data from Process 0 with specific tags. The if size < 4 check also acts as a form of initial process coordination, ensuring the correct environment before communication attempts.
